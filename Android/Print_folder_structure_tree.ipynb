```python
import pathlib

def generate_tree(path, indent="", exclude_list=None):
    if exclude_list is None:
        exclude_list = ["fonts", "font"] # Add any folder names you want to skip here
        
    path = pathlib.Path(path)
    
    try:
        items = sorted(path.iterdir(), key=lambda x: (x.is_file(), x.name.lower()))
    except PermissionError:
        return

    for i, item in enumerate(items):
        is_last = i == len(items) - 1
        connector = "└── " if is_last else "├── "
        
        print(f"{indent}{connector}{item.name}")
        
        # Check if it's a directory AND not in our exclusion list
        if item.is_dir():
            if item.name.lower() in [name.lower() for name in exclude_list]:
                # We skip the recursion but we've already printed the name above
                continue
                
            extension = "    " if is_last else "│   "
            generate_tree(item, indent + extension, exclude_list)

if __name__ == "__main__":
    target_path = r"C:\Users\hatta\Desktop\presonal-projects\Tunis_transport_map\code\Android"
    
    # Define which folders to see but not enter
    to_skip = ["build","fonts", "font", ".git", "node_modules", "__pycache__"]
    
    print(f"Folder PATH listing for volume {pathlib.Path(target_path).anchor}")
    print(target_path)
    generate_tree(target_path, exclude_list=to_skip)
```
